```python
import pandas as pd
df=pd.read_csv("creditcard.csv")
print(df)
```

```python
print("shape of dataset:",df.shape)
print("dataset columns:",df.columns.tolist)
if "Class" not in df.columns:
    raise ValueError("Target column 'Class' is missing from the dataset.")
x=df.drop(columns=['Class'])
y=df['Class']
print("X shape:",x.shape)
print("y shape:", y.shape)
print(x.columns.tolist())
print(y.value_counts().sort_index())
```

```python
print(y.value_counts(normalize=True).sort_index().round(4))
print("\n--- Validation ---")
print("Class in x:", "Class" in x.columns)
print("X and y have matching row counts:", len(x) == len(y))
print("Target contains missing values:", y.isnull().any())
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42, stratify=y
)
print("x_train:", x_train.shape)
print("x_test:", x_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
print(y_train.value_counts().sort_index())
print(y_test.value_counts().sort_index())
y_train.value_counts(normalize=True).sort_index().round(4)

```

```python
from sklearn.preprocessing import StandardScaler
import pandas as pd
print("=== STEP 26: APPLY STANDARD SCALER ===")
scaler=StandardScaler()
scaler.fit(x_train)
print("Mean:", scaler.mean_)
print("Variance:", scaler.var_)
print("Scale:", scaler.scale_)
x_train_scaled_array=scaler.transform(x_train)
x_test_scaled_array=scaler.transform(x_test)

```

```python
print("\n--- Learned Mean ---")
print(scaler.mean_[:5])

print("\n--- Learned Variance ---")
print(scaler.var_[:5])

print("\n--- Learned Scale ---")
print(scaler.scale_[:5])
print("\n--- Scaled Data Shapes ---")
print("\n--- Learned Mean ---")
print(scaler.mean_[:5])

print("\n--- Learned Variance ---")
print(scaler.var_[:5])

print("\n--- Learned Scale ---")
print(scaler.scale_[:5])

print("\n--- Scaled Data Shapes ---")
print("Training:", x_train_scaled_array.shape)
print("Testing:", x_test_scaled_array.shape)

print("\n--- First Five Scaled Training Means ---")
print(x_train_scaled_array.mean(axis=0)[:5].round(4))
print("\n--- First Five Scaled Testing Means ---")
print(x_test_scaled_array.mean(axis=0)[:5].round(4))

# 6. Check training mean
print("\n--- First Five Scaled Training Means ---")
print(x_train_scaled_array.mean(axis=0)[:5].round(4))

# 6. Check training mean
print("\n--- First Five Scaled Training Means ---")
print(x_train_scaled_array.mean(axis=0)[:5].round(4))
```
